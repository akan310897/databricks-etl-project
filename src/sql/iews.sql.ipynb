```python
%sql
-- COMMAND ----------
-- MAGIC %sql
CREATE OR REPLACE VIEW profit_by_year AS
SELECT year, SUM(total_profit) AS total_profit
FROM retail_catalog.retail.gold_sales
GROUP BY year;
```

```python
%sql
SELECT * from profit_by_year
```

```python
%sql
-- COMMAND ----------
-- MAGIC %sql
CREATE OR REPLACE VIEW profit_by_year_cat AS
SELECT year, SUM(total_profit) AS total_profit
FROM retail_catalog.retail.gold_sales
GROUP BY year, category;
```

```python
%sql
SELECT * from profit_by_year_cat
```

```python
%sql
-- COMMAND ----------
-- MAGIC %sql
CREATE OR REPLACE VIEW profit_by_cust_id AS
SELECT customer_id,SUM(total_profit) AS total_profit
FROM retail_catalog.retail.gold_sales
GROUP BY customer_id;
```

```python
%sql
SELECT * from profit_by_cust_id
```

```python
%sql
-- COMMAND ----------
-- MAGIC %sql
CREATE OR REPLACE VIEW profit_by_cust_year AS
SELECT customer_id,year, SUM(total_profit) AS total_profit
FROM retail_catalog.retail.gold_sales
GROUP BY customer_id,year;
```

```python
%sql
SELECT * from profit_by_cust_year
```
